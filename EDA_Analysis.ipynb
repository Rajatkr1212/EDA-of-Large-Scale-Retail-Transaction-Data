# **--- Outlier Detection ---**

**1. Boxplots**

```python
#Checking all Numerical columns

for col in numeric_columns:
  plt.figure(figsize=(10, 3))
  sns.boxplot(x=df[col], color="orange")
  plt.title(f'Boxplot of {col}')
  plt.xlabel(col)
  plt.tight_layout()
  plt.show()
```

**2. Interquartile Range (IQR) Method**

```python
print("\n" + "=" * 70)
print("OUTLIER DETECTION USING IQR")
print("=" * 70)

def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    return Q1, Q3, IQR, lower_limit, upper_limit, outliers


outlier_columns = [
    "age",
    "product_price",
    "quantity",
    "discount_percentage",
    "final_price",
    "delivery_days"
]

outlier_results = []

for column in outlier_columns:

    Q1, Q3, IQR, lower, upper, outliers = detect_outliers_iqr(
        df, column
    )

    outlier_results.append({
        "Variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower,
        "Upper Limit": upper,
        "Number of Outliers": len(outliers),
        "Outlier Percentage": round(
            len(outliers) / len(df) * 100, 2
        )
    })


outlier_df = pd.DataFrame(outlier_results)

display(outlier_df)
```

**MULTIVARIATE ANALYSIS**

```python
#Product Category × Discount × Final Price
df["discount_group"] = pd.cut(
df["discount_percentage"],
bins=[-1, 10, 20, 30],
labels=[
"0-10%",
"11-20%",
"21-30%"
]
)

category_discount_price = (
df.groupby(
[
"product_category",
"discount_group"
],
observed=True
)["final_price"]
.mean()
.reset_index()
)

print("\nProduct Category × Discount × Final Price:")
print(
category_discount_price
.sort_values("final_price", ascending=False)
.round(2)
)

plt.figure(figsize=(12, 6))

sns.barplot(
data=category_discount_price,
x="product_category",
y="final_price",
hue="discount_group"
)

plt.title(
"Product Category × Discount Group × Average Final Price"
)

plt.xlabel("Product Category")
plt.ylabel("Average Final Price")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
print("\n")

plt.figure(figsize=(12, 6))

sns.boxplot(
data=df,
x="product_category",
y="discount_percentage"
)
```

```python
#CUSTOMER SEGMENT × AGE × SPENDING
df["age_group"] = pd.cut(
df["age"],
bins=[17, 30, 45, 60, 100],
labels=[
"18-30",
"31-45",
"46-60",
"61+"
]
)

age_segment_spending = (
df.groupby(
[
"customer_segment",
"age_group"
],
observed=True
)["final_price"]
.mean()
.reset_index()
)

print("\nCustomer Segment × Age × Final Price:")
print(
age_segment_spending
.sort_values("final_price", ascending=False)
.round(2)
)

pivot_spending = age_segment_spending.pivot(
index="customer_segment",
columns="age_group",
values="final_price"
)

plt.figure(figsize=(10, 5))

sns.heatmap(
pivot_spending,
annot=True,
fmt=".2f",
cmap="YlGnBu"
)

plt.title(
"Average Final Price by Customer Segment and Age Group"
)

plt.xlabel("Age Group")
plt.ylabel("Customer Segment")

plt.tight_layout()
plt.show()
print("\n")

plt.figure(figsize=(10, 5))

sns.barplot(
data=age_segment_spending,
x="customer_segment",
y="final_price",
hue="age_group"
)
```

```python
#DELIVERY DAYS × SHIPPING TYPE × RETURN STATUS
shipping_return = (
df.groupby("shipping_type")["return_encoded"]
.mean()
.mul(100)
)

print("\nReturn Rate by Shipping Type:")
print(shipping_return.round(2))

df["delivery_group"] = pd.cut(
df["delivery_days"],
bins=[0, 3, 6, 10],
labels=[
"1-3 Days",
"4-6 Days",
"7-10 Days"
]
)

delivery_return_analysis = (
df.groupby(
[
"shipping_type",
"delivery_group"
],
observed=True
)["return_encoded"]
.mean()
.mul(100)
.reset_index(name="return_rate")
)

print(
"\nDelivery Days × Shipping Type × Return Rate:"
)

print(
delivery_return_analysis
.sort_values("return_rate", ascending=False)
.round(2)
)

plt.figure(figsize=(8, 4))

sns.barplot(
data=delivery_return_analysis,
x="delivery_group",
y="return_rate",
hue="shipping_type"
)

plt.title(
"Return Rate by Delivery Duration and Shipping Type"
)

plt.xlabel("Delivery Duration")
plt.ylabel("Return Rate (%)")

plt.tight_layout()
plt.show()
print("\n")

plt.figure(figsize=(8, 4))

sns.boxplot(
data=df,
x="shipping_type",
y="delivery_days")
```

**Skewness and Kurtosis Interpretation**


```python
#SKEWNESS ANALYSIS
print("\n" + "=" * 70)
print("SKEWNESS ANALYSIS")
print("=" * 70)

skewness = (
df[numeric_columns]
.skew()
.sort_values(ascending=False)
)

print(skewness.round(4))

plt.figure(figsize=(8, 4))

sns.boxplot(
data=df,
y="final_price",
orient="h"
)
```

```python
#KURTOSIS ANALYSIS
print("\n" + "=" * 70)
print("KURTOSIS ANALYSIS")
print("=" * 70)

kurtosis = (
df[numeric_columns]
.kurtosis()
.sort_values(ascending=False)
)

print(kurtosis.round(4))

plt.figure(figsize=(10, 6))

sns.histplot(
data=df,
x="final_price",
kde=True)
```
